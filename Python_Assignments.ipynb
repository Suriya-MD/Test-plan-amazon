```python
'''1.	Print all prime numbers between input range (Ex – input 20 50, prints all prime numbers between 20 and 50).'''

start=int(input())
end=int(input())
for num in range(start,end+1):
  count=0
  for i in range(1,num+1):
    if num%i==0:
      count+=1
  if count==2:
    print(num,end=" ")
```

```python
'''2.	Factorial using recursion'''

def fact(n):
  if n==0 or n==1:
    return 1
  else:
    return n*fact(n-1)

n=int(input())
print(fact(n))
```

```python
'''3.	Square of numbers using lambda'''

n=int(input())
square= lambda x: x*x
print(square(n))
```

```python
'''4.	Find the second largest element in a list'''

def sec_lar(nums):
  lar=max(nums[0],nums[1])
  sec=min(nums[0],nums[1])
  for x in nums[2:]:
    if x>lar:
      sec=lar
      lar=x
    elif x>sec:
      sec=x
  return sec

nums=list(map(int,input().split()))
print(sec_lar(nums))
```

```python
'''5.	Count frequency of characters in a string'''

str=input()
freq={}
for ch in str:
  if ch in freq:
    freq[ch]+=1
  else:
    freq[ch]=1
print(freq)
```

```python
'''6.	Calculate area of a circle using math library.'''

import math
r=float(input())
print("Area of circle",f"{math.pi*r*r:.2f}")
```

```python
'''7.	Reverse a string without using built‑in reverse'''

str=input()
rev=""
for x in str:
  rev=x+rev
print(rev)
```

```python
'''8.	Remove duplicates from a list'''

n=list(map(int,input().split()))
new=[]
for i in n:
  if i not in new:
    new.append(i)

print(new)
```

```python
'''9.	Merge two dictionaries'''

a = {"a": 1, "b": 2}
b = {"c": 3, "d": 4}
a.update(b)
print(a)
```

```python
'''10.	Fibonacci series using recursion'''

def fibo(n):
  if n<=1:
    return n
  else:
    return fibo(n-1)+fibo(n-2)

n=int(input())
for i in range(n):
  print(fibo(i),end=" ")
```
