```python
'''1.	Write a function calculate(a, b, operation) that performs addition,
subtraction, multiplication, or division based on the supplied operation.'''

def calculate(a,b,op):
  if op == "+":
        return a + b
  elif op == "-":
      return a - b
  elif op == "*":
      return a * b
  elif op == "/":
      return a / b

a,b=map(int,input().split())
op=input()
print(calculate(a,b,op))
```

```python
'''2.	Write a function sum_numbers(*args) that accepts any number of
arguments and returns their sum.'''

def sum_numbers(*args):
  sum=0
  for i in args:
    sum+=i
  return sum

nums=list(map(int,input().split()))
print(sum_numbers(*nums))

```

```python
'''3.	Write a function employee(**args) that accepts employee information such as name, ID, department and salary,
then displays the information.'''

def employee(**args):
  for key,values in args.items():
    print(key," : ",values)

name=input()
id=int(input())
dept=input()
sal=int(input())
employee(Name=name,ID=id,Dept=dept,Salary=sal)

```

```python
'''4.	Write a function remove_duplicates(lst) that returns a list containing only
unique elements while preserving their original order.'''

def remove_duplicates(lst):
    result = []
    for x in lst:
        if x not in result:
            result.append(x)
    return result

lst = list(map(int, input().split()))
print(remove_duplicates(lst))
```

```python
'''5.	Using a lambda function, sort a list of tuples based on the second element.
 Example: [(1,5), (2,3), (4,1)].'''

lst = [(1, 5), (2, 3), (4, 1)]
res=sorted(lst,key=lambda x: x[1])
print(res)
```
